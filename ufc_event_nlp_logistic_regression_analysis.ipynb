```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Text preprocessing
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    return text.strip()

# Example training data
texts = [
    "fighter dominated striking and won by knockout",
    "fighter secured takedowns and won by decision",
    "fighter was knocked out early",
    "fighter controlled grappling and submitted opponent"
]

labels = ["fighter1", "fighter1", "fighter2", "fighter1"]

# Preprocess training data
clean_texts = [preprocess_text(t) for t in texts]

# Create and fit vectorizer
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(clean_texts)

# Train model
model = LogisticRegression()
model.fit(X, labels)

# New fight prediction
new_fight = ["fighter dominated striking and finished with knockout"]

clean = preprocess_text(new_fight[0])
vec = vectorizer.transform([clean])

prediction = model.predict(vec)

print("Predicted Winner:", prediction[0])
```
